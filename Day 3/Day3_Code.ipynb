```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('household_power_consumption_cleaned.csv', sep=';')
```

```python
numeric_cols = ['Global_active_power','Global_reactive_power','Voltage', 
                'Global_intensity','Sub_metering_1','Sub_metering_2','Sub_metering_3']
```

```python
df[numeric_cols] = df[numeric_cols].astype(float)
```

```python
df
```

```python
df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')
```

```python
df
```

```python
df['hour'] = df['datetime'].dt.hour
df['day_of_week'] = df['datetime'].dt.dayofweek
```

```python
df['hour']
```

```python
df['day_of_week']
```

```python
df['hour_sin'] = np.sin(2 * np.pi * df['hour']/24)
df['hour_cos'] = np.cos(2 * np.pi * df['hour']/24)
```

```python
df['hour_sin'] 
```

```python
df['hour_cos']
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
scaler_std = StandardScaler()
scaler_minmax = MinMaxScaler()
```

```python
df_std = pd.DataFrame(scaler_std.fit_transform(df[numeric_cols]), columns=numeric_cols)
```

```python
df_std
```

```python
df_minmax = pd.DataFrame(scaler_minmax.fit_transform(df[numeric_cols]), columns=numeric_cols)
```

```python
df_minmax
```

```python
df['Apparent_Power'] = (df['Global_active_power']**2 + df['Global_reactive_power']**2)**0.5
```

```python
df['Apparent_Power']
```

```python
df.to_csv('features.csv', index=False)
```
